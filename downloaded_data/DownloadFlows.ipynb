```python
import sys
import shutil
import pexpect
import getpass

if sys.platform == "win32":
    from pexpect.popen_spawn import PopenSpawn
```

```python
def DownloadFile(file_name, password=None):
  remote = f"dennyh23@aina.ifa.hawaii.edu:/SCIPE/{file_name}"
  if sys.platform == "win32":
    scp = shutil.which("scp") or r"C:\Windows\System32\OpenSSH\scp.exe"
    child = PopenSpawn([scp, remote, "."], encoding="utf-8")
  else:
    child = pexpect.spawn(f"scp {remote} .")
  # Pass a list of possible prompts
  # 0 = host key verification, 1 = password prompt
  index = child.expect(["Are you sure you want to continue connecting", "password:"])
  if index == 0:
    # Handle the authenticity warning by sending 'yes'
    child.sendline("yes")
    # Now wait for the password prompt that follows
    child.expect("password:")

    if password is None:
      # Prompt the user while replacing characters with '*'
      #password = getpass.getpass("Enter your password: ", echo_char="*")
      password = getpass.getpass("Enter your password: ")
    child.sendline(password)

  elif index == 1:
    if password is None:
      # The host was already known, went straight to password
      password = getpass.getpass("Enter your password: ")
    child.sendline(password)

  # Wait for the transfer to finish and print the output
  child.expect(pexpect.EOF, timeout=180)
  output = child.before
  if isinstance(output, bytes):
    output = output.decode()
  print(output)
  return password
```
