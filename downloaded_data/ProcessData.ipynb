```python
from pathlib import Path
import zipfile
import shutil
import pandas as pd
```

```python
# Folder containing your ZIP files
zip_folder = Path("data")
```

```python
# Where you want all CSVs to go
csv_folder = Path("processed_data")
csv_folder.mkdir(parents=True, exist_ok=True)
```

```python
for zip_path in zip_folder.rglob("*.zip"):
    with zipfile.ZipFile(zip_path, "r") as z:
        for file_info in z.infolist():
            if file_info.is_dir():
                continue
            # Only process CSV files
            if not file_info.filename.lower().endswith(".csv"):
                continue

            # Get just the CSV filename, ignoring folders inside the ZIP
            csv_name = Path(file_info.filename).name

            # Destination path
            output_path = csv_folder / csv_name

            # Extract the CSV
            with z.open(file_info) as source, open(output_path, "wb") as target:
                shutil.copyfileobj(source, target)

            print(f"Extracted: {zip_path} -> {output_path}")
```

```python
files = sorted(csv_folder.rglob("*.csv"))

# Write first file with header, then append the rest
pd.read_csv(files[0]).to_csv("combined.csv", index=False, mode="w")
for f in files[1:]:
    pd.read_csv(f).to_csv("combined.csv", index=False, header=False, mode="a")
```
